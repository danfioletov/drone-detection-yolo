```python
from ultralytics import YOLO
import os
import pandas as pd
A = "/content/drive/MyDrive/Computer Vision/yolo_runs/drone_highres3/weights/best.pt"                         # ← вкажіть, де лежить best3.pt
B = "/content/drive/MyDrive/Computer Vision/yolo_runs_shahed/drone_shahed4/weights/best.pt"
for p in (A, B):
    print(os.path.exists(p), p)


ROOT = "/content/drive/MyDrive/Computer Vision"
MODELS = {
    "A_kaggle_highres3": f"{ROOT}/yolo_runs/drone_highres3/weights/best.pt",
    "B_synth_shahed4":   f"{ROOT}/yolo_runs_shahed/drone_shahed4/weights/best.pt",
}
YAML = f"{ROOT}/datasets/real_shahed/real_test.yaml"     # ← ваш фактичний шлях
OUT  = f"{ROOT}/yolo_runs_shahed/eval_real"

rows = {}
for name, weights in MODELS.items():
    print(f"\n######## {name}")
    model = YOLO(weights)
    print("класи:", model.names)
    m = model.val(data=YAML, split="test", imgsz=640,
                  project=OUT, name=name, exist_ok=True)
    rows[name] = {
        "P":        m.box.mp,
        "R":        m.box.mr,
        "mAP50":    m.box.map50,
        "mAP50-95": m.box.map,
    }

pd.DataFrame(rows).T.round(3)
```
